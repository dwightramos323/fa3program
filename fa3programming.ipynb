```python
#fa3prog1
#Write a program that:

#a. Asks for the student's name

first_name = None
last_name = None

while not first_name:
    first_name = input("What is your first name? ").upper()

while not last_name:
    last_name = input("What is your last name? ").upper()

#b. Asks for the student's age

age = None
while age is None:
    try:
        age = int(input("What is your age? "))
    except ValueError:
        print("Please enter a valid age.")

#c. Asks for the student's course

course = None
course_list = ['SS', 'IS', 'MATH', 'CS', 'ENGLISH', 'FILIPINO', 'ADTECH', 'PE', 'HEALTH', 'MUSIC', 'VALED']

while course not in course_list:
    print(course_list)
    course = input("What is your course? (choose from the list): ").upper()
    if course not in course_list:
        print("Please choose a valid course from the list.")

#d. Displays all the entered information in a formatted output

print(f"\nStudent Name: {first_name} {last_name}")
print(f"Student Age: {age}")
print(f"Student Course: {course}")
```

```python
# fa3prog2
# Pass or Fail Checker

score = None

# Ask for score with validation
while score is None:
    try:
        score = int(input("Enter student's exam score: "))
    except ValueError:
        print("Please enter a valid number.")

# Check if passed or failed
if score >= 75:
    print("Passed")
else:
    print("Failed")
```

```python
# fa3prog3
# Grading System

# Ask for score with validation
score = None
while score is None:
    try:
        score = int(input("Enter student's score: "))
    except ValueError:
        print("Please enter a valid number.")

# Determine grade
if 90 <= score <= 100:
    grade = "A"
elif 80 <= score <= 89:
    grade = "B"
elif 70 <= score <= 79:
    grade = "C"
else:
    grade = "F"

# Display result
print("\n--- Grading Result ---")
print(f"Score: {score}")
print(f"Grade: {grade}")
```
